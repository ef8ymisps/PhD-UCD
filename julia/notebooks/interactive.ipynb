```julia
function audit_powermodels_units(network_data)
    println("name = ", get(network_data, "name", "unknown"))
    println("baseMVA = ", network_data["baseMVA"])
    println("n_bus = ", length(network_data["bus"]))
    println("n_load = ", length(network_data["load"]))
    println("n_gen = ", length(network_data["gen"]))
    println("n_branch = ", length(network_data["branch"]))

    println("\nGenerators:")
    for gen_id in sorted_keys_by_int(network_data["gen"])
        gen = network_data["gen"][gen_id]
        println(
            "gen ", gen_id,
            " pg=", gen["pg"],
            " pmin=", gen["pmin"],
            " pmax=", gen["pmax"],
            " cost=", get(gen, "cost", missing)
        )
    end

    println("\nFirst loads:")
    for load_id in first(sorted_keys_by_int(network_data["load"]), min(10, length(network_data["load"])))
        load = network_data["load"][load_id]
        println(
            "load ", load_id,
            " bus=", load["load_bus"],
            " pd=", load["pd"],
            " qd=", load["qd"]
        )
    end

    println("\nFirst branches:")
    for branch_id in first(sorted_keys_by_int(network_data["branch"]), min(10, length(network_data["branch"])))
        br = network_data["branch"][branch_id]
        println(
            "branch ", branch_id,
            " r=", br["br_r"],
            " x=", br["br_x"],
            " rate_a=", br["rate_a"],
            " angmin=", get(br, "angmin", missing),
            " angmax=", get(br, "angmax", missing),
            " tap=", get(br, "tap", missing),
            " shift=", get(br, "shift", missing)
        )
    end
end
audit_powermodels_units(network_data)
```

```julia
using Statistics
using PowerModels
using HiGHS
using Ipopt

include("C:/Users/Efthymis/OneDrive - University College Dublin/power-systems-tools-learning/julia/own_models/src/OwnPowerModels.jl")
using .OwnPowerModels

root_dir = "C:/Users/Efthymis/OneDrive - University College Dublin/power-systems-tools-learning"
case_name = "pglib_opf_case14_ieee.m"
case_path = joinpath(root_dir, "data", "raw", "pglib-opf", case_name)

case = OwnPowerModels.load_powercase_from_powermodels(case_path)

highs_silent = optimizer_with_attributes(HiGHS.Optimizer, "output_flag" => false)
ipopt_silent = optimizer_with_attributes(Ipopt.Optimizer, "print_level" => 0)

function benchmark_call(label, f; runs=5)
    println("\n", label)
    println("warming up...")
    f()

    times = Float64[]

    for k in 1:runs
        GC.gc()
        t = @elapsed f()
        push!(times, t)
        println("run ", k, ": ", round(t, digits=4), " s")
    end

    println("min:    ", round(minimum(times), digits=4), " s")
    println("median: ", round(median(times), digits=4), " s")
    println("mean:   ", round(mean(times), digits=4), " s")

    return times
end

println("\n==============================")
println("DC OPF SPEED COMPARISON")
println("==============================")

own_dc_times = benchmark_call("Own DC OPF solve only", runs=10) do
    OwnPowerModels.solve_dc_opf_own(case)
end

pm_dc_times = benchmark_call("PowerModels DC OPF end-to-end", runs=10) do
    PowerModels.solve_dc_opf(case_path, highs_silent)
end

own_dc_e2e_times = benchmark_call("Own DC OPF end-to-end: parse + solve", runs=10) do
    local c = OwnPowerModels.load_powercase_from_powermodels(case_path)
    OwnPowerModels.solve_dc_opf_own(c)
end

println("\n==============================")
println("AC OPF SPEED COMPARISON")
println("==============================")

own_ac_times = benchmark_call("Own AC OPF solve only", runs=5) do
    OwnPowerModels.solve_ac_opf_own(case)
end

pm_ac_times = benchmark_call("PowerModels AC OPF end-to-end", runs=5) do
    PowerModels.solve_ac_opf(case_path, ipopt_silent)
end

own_ac_e2e_times = benchmark_call("Own AC OPF end-to-end: parse + solve", runs=5) do
    local c = OwnPowerModels.load_powercase_from_powermodels(case_path)
    OwnPowerModels.solve_ac_opf_own(c)
end
```
